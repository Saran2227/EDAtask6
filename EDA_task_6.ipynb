```python
df.groupby(["Medical_Condition", "Admission_Type"])["Billing_Amount"].mean().unstack().plot(kind="bar",
figsize=(10, 5))
plt.title("Average Billing Amount by Medical Condition & Admission Type");
plt.ylabel("Billing Amount");
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
billing_summary=(
    df.groupby(['Medical_Condition', 'Admission_Type'])['Billing_Amount']
    .sum()
    .unstack()
)
billing_summary.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    edgecolor='black'
)

plt.title('Total Billing Amount by Medical Condition and Admission Type',
          fontsize=16)

plt.xlabel('Medical Conditon', fontsize=12)
plt.ylabel('Total Billing Amount(₹)', fontsize=12)

plt.xticks(rotation=45)
plt.legend(title='Admission Type')

plt.show()


```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df,
    x='Medical_Condition',
    y='Billing_Amount'
)

plt.title('Billing Amount Distribution by Medical Condition',
          fontsize=16)

plt.xlabel('Medical Condition')
plt.ylabel('Billing Amount(₹)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

```

```python
df['Date_of_Admission'] = pd.to_datetime(
    df['Date_of_Admission']
)
```

```python
monthly_admissions = (
    df.set_index('Date_of_Admission')
      .resample('ME')
      .size()
)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_admissions.index,
    monthly_admissions.values,
    marker='o'
)

plt.title('Monthly Patient Admissions', fontsize=16)
plt.xlabel('Montly', fontsize=12)
plt.ylabel('Number of Admissions', fontsize=12)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
columns_to_check = ['Age', 'Stay_Days', 'Billing_Amount']
correlation_subset = df[columns_to_check].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(
    correlation_subset,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
)
plt.title('Correlation: Age, Stay Days & Billing Amount', fontsize=20)
plt.show()
```
